# Get max curated data

```python
import requests
from pathlib import Path
from yaml import safe_load
import numpy as np
import re

DATA_DIR = Path("data/max-curated")
DATA_DIR.mkdir(exist_ok=True, parents=True)
FIGURES_DIR = Path("figures")
FIGURES_DIR.mkdir(exist_ok=True, parents=True)


def figure_output_path(filename):
    return FIGURES_DIR / f"case-1-{filename}"


CHEMBL_VERSION = "32"  # chembl 32 was used in the IC50 comparability paper

activities = ["Ki", "IC50"]
data_url = "https://raw.githubusercontent.com/rinikerlab/overlapping_assays/refs/heads/main/datasets/{dataset}_datasets.yaml"


command_str = (
    "capricho get -tids {targets} -chiral -duchi -reqdoc "
    "-cure --max-assay-size 100 --min-assay-size 20 --min-assay-overlap 5 -smr -calc -c 9 -idcols assay_type,assay_cell_type "
    "--chembl-version {version} -biotype {activity} -o {output}"
)
```

## Ki

```python
result = requests.get(data_url.format(dataset=activities[0]))
result.raise_for_status()

datasets = safe_load(result.content)

ki_targets = []
keys = datasets["sources"].keys()
for key in keys:
    ki_targets.append(datasets["sources"][key]["description"])

ki_targets = np.unique(ki_targets).tolist()
ki_targets = [re.compile("(CHEMBL\d+)").search(t).group(1) for t in ki_targets]

ki_output = DATA_DIR / f"curated-Ki-NoAssayOverlap-chembl{CHEMBL_VERSION}.csv"
command = command_str.format(
    targets=",".join(ki_targets),
    activity=activities[0],
    output=str(ki_output),
    version=CHEMBL_VERSION,
)

if not ki_output.exists():
    ! {command}
else:
    print(f"Data already exists: {ki_output}")
    print("Delete this file to re-fetch with updated parameters.")
```

## IC50

```python
result = requests.get(data_url.format(dataset=activities[1]))
result.raise_for_status()

datasets = safe_load(result.content)

ic50_targets = []
keys = datasets["sources"].keys()
for key in keys:
    ic50_targets.append(datasets["sources"][key]["description"])

ic50_targets = np.unique(ic50_targets).tolist()
ic50_targets = [re.compile("(CHEMBL\d+)").search(t).group(1) for t in ic50_targets]

ic50_output = DATA_DIR / f"curated-IC50-NoAssayOverlap-chembl{CHEMBL_VERSION}.csv"
command = command_str.format(
    targets=",".join(ic50_targets),
    activity=activities[1],
    output=str(ic50_output),
    version=CHEMBL_VERSION,
)

if not ic50_output.exists():
    ! {command}
else:
    print(f"Data already exists: {ic50_output}")
    print("Delete this file to re-fetch with updated parameters.")
```

# IC50 overlap

Notice that we assign a `source_datapoint_id` to each activity that has aggregated values (the ones having the special pipe character in the processed_smiles column). This allows us to track the original source datapoint for each aggregated value, since we evalue in the notebook the cross-assay comparability.

An example, for the readout `9.5|9.0|8.5`, the scatter plots will have:
- 9.5 v.s. 9.0
- 9.5 v.s. 8.5
- 9.0 v.s. 8.5

Having a `source_datapoint_id` allows us to track the original datapoint for each of these comparisons (_all activity pairs will carry the same source identifier_), which is important in case you're interested in a specific pool of aggregated values.

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

from Capricho.analysis import (
    DroppingComment,
    ProcessingComment,
    explode_assay_comparability,
    get_all_comments,
    plot_cross_assay_coverage,
    plot_multi_panel_comparability,
    plot_subset,
)
```

```python
df = pd.read_csv(ic50_output, engine="pyarrow")
subset = df.query('processed_smiles.str.contains("|", regex=False)').assign(
    source_datapoint_id=lambda x: range(len(x))
)
subset.shape
```

```python
df.data_processing_comment.apply(
    lambda x: x.split("|") if isinstance(x, str) else []
).explode().value_counts()
```

```python
all_comments = get_all_comments()
ic50_cross_assay_comparisons = explode_assay_comparability(subset)
```

```python
fig, axs = plot_multi_panel_comparability(
    ic50_cross_assay_comparisons,
    all_comments,
    title=f"IC50 Comparability Across Flagged Data (ChEMBL {CHEMBL_VERSION})",
    figsize=(22, 9),
    ncols=5,
)
fig.savefig(
    figure_output_path(f"ic50_comparability_allflags_chembl{CHEMBL_VERSION}.png"),
    dpi=300,
    bbox_inches="tight",
    transparent=False,
)
```

```python
drop_flags_rm = [
    DroppingComment.DATA_VALIDITY_COMMENT.value,
    DroppingComment.POTENTIAL_DUPLICATE.value,
]

drop_flags_rm = "|".join(drop_flags_rm)

fig, ax = plot_subset(
    ic50_cross_assay_comparisons.query("(~dropping_comment.str.contains(@drop_flags_rm, regex=True))").query(
        "pchembl_value_x != pchembl_value_y"
    ),
    title="Unprocessed IC50 Data",
    figsize=(5, 4.6),
)

fig.savefig(
    figure_output_path(f"ic50_unprocessed_comparability_chembl{CHEMBL_VERSION}.png"),
    dpi=300,
    bbox_inches="tight",
    transparent=False,
)
```

```python
drop_flags_rm = [
    DroppingComment.DATA_VALIDITY_COMMENT.value,
    DroppingComment.POTENTIAL_DUPLICATE.value,
    DroppingComment.ASSAY_SIZE_TOO_SMALL.value,
    DroppingComment.MUTATION_KEYWORD.value,
    DroppingComment.UNDEFINED_STEREOCHEMISTRY.value,
    DroppingComment.UNIT_ANNOTATION_ERROR.value,
]
rm_process_obs = ProcessingComment.PCHEMBL_DUPLICATION_ACROSS_DOCUMENTS.value

drop_flags_rm = "|".join(drop_flags_rm)

subset_ic50_clean = ic50_cross_assay_comparisons.query(
    "(~dropping_comment.str.contains(@drop_flags_rm, regex=True)) "
    "& ~processing_comment.str.contains(@rm_process_obs, regex=True)"
    "& pchembl_value_x != pchembl_value_y"
)
fig, ax = plot_subset(subset_ic50_clean, title="Cleaned IC50 Data", figsize=(5, 4.6))
fig.savefig(
    figure_output_path(f"ic50_clean_comparability_chembl{CHEMBL_VERSION}.png"),
    dpi=300,
    bbox_inches="tight",
    transparent=False,
)
```

# Data omitted by quality filtering

Cross-assay agreement must be interpreted together with the amount of data available.

The table therefore reports the selected quality flags at three explicitly labelled levels: retrieved measurements, aggregated compound–target datapoints, and pairwise comparisons. `Any of the selected flags` is the union of the reasons and is the absolute and percentage amount omitted by the corresponding cleaned panel.

```python
from Capricho.flag_report import (
    coverage_table,
    format_coverage_table,
    format_curation_summary,
    summarize_curation,
)


def not_aggregated(path):
    return pd.read_csv(path.with_name(path.stem + "_not_aggregated.csv"), engine="pyarrow")


case1_datasets = [
    (
        f"IC50 (ChEMBL {CHEMBL_VERSION})",
        not_aggregated(ic50_output),
        df,
        ic50_cross_assay_comparisons,
        [
            DroppingComment.DATA_VALIDITY_COMMENT.value,
            DroppingComment.POTENTIAL_DUPLICATE.value,
            DroppingComment.ASSAY_SIZE_TOO_SMALL.value,
            DroppingComment.MUTATION_KEYWORD.value,
            DroppingComment.UNDEFINED_STEREOCHEMISTRY.value,
            DroppingComment.UNIT_ANNOTATION_ERROR.value,
        ],
    ),
    (
        f"Ki (ChEMBL {CHEMBL_VERSION})",
        not_aggregated(ki_output),
        df_ki,
        ki_cross_assay_comparisons,
        [
            DroppingComment.DATA_VALIDITY_COMMENT.value,
            DroppingComment.POTENTIAL_DUPLICATE.value,
            DroppingComment.ASSAY_SIZE_TOO_LARGE.value,
            DroppingComment.ASSAY_SIZE_TOO_SMALL.value,
        ],
    ),
]

case1_omission = pd.concat(
    [
        summarize_curation(name, retrieved, aggregated, comparisons, drop_flags)
        for name, retrieved, aggregated, comparisons, drop_flags in case1_datasets
    ],
    ignore_index=True,
)
print(format_curation_summary(case1_omission))
case1_omission.to_csv(f"case1_omission_summary_chembl{CHEMBL_VERSION}.csv", index=False)
```

## Cross-assay coverage

To get a sense of how much of the data can be used for checking the cross-assay comparability, we report the comparability coverage at two levels:

1. The number and percentage of datapoints (compound-target affinity) that are reported more than once in the dataset, measured by either the same by different assays.
2. The number and percentage of assays that share at least one datapoint with another assay.

Although the cross-assay comparability can be a proxy for estimating whether the readouts are comparable for aggregation; the coverage of these overlaps should be considered when interpreting the results. E.g., if only a small fraction of the datapoints are measured in more than one assay, the cross-assay comparability may not be representative of the entire dataset.
