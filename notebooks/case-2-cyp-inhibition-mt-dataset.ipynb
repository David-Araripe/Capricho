## Fetch CYP data

To fetch the data, we use the CLI with the following parameters (`-v` selects the ChEMBL
release; set the `CHEMBL_VERSION` variable below to change it):
```bash
capricho get --target-ids CHEMBL340,CHEMBL289,CHEMBL3397,CHEMBL3622,CHEMBL3356,CHEMBL2722,CHEMBL1908,CHEMBL3379,CHEMBL3978,CHEMBL4878,CHEMBL2231,CHEMBL3721,CHEMBL5282,CHEMBL4729 \
    --drop-unassigned-chiral --chirality --min-assay-size 5 --max-assay-size 100 \
    --assay-types A,B,F --id-columns standard_type --output-path cyp_chembl_data.csv -v 36
```

````python
from pathlib import Path

DATA_DIR = Path("data/cyp")
DATA_DIR.mkdir(exist_ok=True, parents=True)
FIGURES_DIR = Path("figures")
FIGURES_DIR.mkdir(exist_ok=True, parents=True)


def figure_output_path(filename):
    return FIGURES_DIR / f"case-2-{filename}"


# The manuscript reports the CYP results on ChEMBL 36. Set this to a newer release (e.g.
# "37") and re-run to repeat the analysis on current data; the output CSV and figures are
# suffixed with the version, so runs do not overwrite each other.
CHEMBL_VERSION = "36"

DATA_FILE = DATA_DIR / f"cyp_chembl_data-chembl{CHEMBL_VERSION}.csv"

if not DATA_FILE.exists():
    ! capricho get --target-ids CHEMBL340,CHEMBL289,CHEMBL3397,CHEMBL3622,CHEMBL3356,CHEMBL2722,CHEMBL1908,CHEMBL3379,CHEMBL3978,CHEMBL4878,CHEMBL2231,CHEMBL3721,CHEMBL5282,CHEMBL4729 \
        --drop-unassigned-chiral --chirality --min-assay-size 5 --max-assay-size 100 \
        --assay-types A,B,F --id-columns standard_type --output-path {str(DATA_FILE)} -v {CHEMBL_VERSION}
else:
    print(f"Data already exists: {DATA_FILE}")
    print("Delete this file to re-fetch with updated parameters.")
````

## Data omitted by quality filtering

Cross-assay agreement must be interpreted together with the amount of data available.

The table therefore reports the selected quality flags at three explicitly labelled levels: retrieved measurements, aggregated compound–target datapoints, and pairwise comparisons. `Any of the selected flags` is the union of the reasons and is the absolute and percentage amount omitted by the corresponding cleaned panel.

````python
from Capricho.flag_report import format_curation_summary, summarize_curation

CLEANED_PANEL_FLAGS = [
    DroppingComment.UNIT_ANNOTATION_ERROR.value,
    DroppingComment.UNDEFINED_STEREOCHEMISTRY.value,
    DroppingComment.ASSAY_SIZE_TOO_SMALL.value,
]
cyp_retrieved = pd.read_csv(DATA_FILE.with_name(DATA_FILE.stem + "_not_aggregated.csv"), engine="pyarrow")
cyp_omission = summarize_curation(
    f"CYP (ChEMBL {CHEMBL_VERSION})",
    cyp_retrieved,
    df,
    cyp_cross_assay_comparisons,
    CLEANED_PANEL_FLAGS,
)
print(format_curation_summary(cyp_omission))
cyp_omission.to_csv(f"cyp_omission_summary_chembl{CHEMBL_VERSION}.csv", index=False)
````

## Cross-assay coverage per isoform

To get a sense of how much of the data can be used for checking the cross-assay comparability, we report the comparability coverage at two levels:

1. The number and percentage of datapoints (compound-target affinity) that are reported more than once in the dataset, measured by either the same by different assays.
2. The number and percentage of assays that share at least one datapoint with another assay.

Although the cross-assay comparability can be a proxy for estimating whether the readouts are comparable for aggregation; the coverage of these overlaps should be considered when interpreting the results. E.g., if only a small fraction of the datapoints are measured in more than one assay, the cross-assay comparability may not be representative of the entire dataset.

_For CYP, we report these values per isoform, as well as for the entire panel_:

````python
from Capricho.chembl.api.downloader import get_target_names_sql
from Capricho.flag_report import coverage_table, format_coverage_table, summarize_curation_by_group

isoform_names = {
    target_id: name.replace("Cytochrome P450 ", "CYP ").replace(", mitochondrial", "")
    for target_id, name in get_target_names_sql(
        sorted(df["target_chembl_id"].unique()), version=CHEMBL_VERSION
    ).items()
}

cyp_overall_coverage = coverage_table(cyp_omission)

cyp_coverage = coverage_table(
    summarize_curation_by_group(
        "target_chembl_id",
        cyp_retrieved,
        df,
        cyp_cross_assay_comparisons,
        CLEANED_PANEL_FLAGS,
        labels=isoform_names,
    )
).sort_values("datapoint_overlap_pct", ascending=False, ignore_index=True)

cyp_coverage.to_csv(f"cyp_coverage_by_isoform_chembl{CHEMBL_VERSION}.csv", index=False)

display(format_coverage_table(cyp_coverage, dataset_label="Isoform"))
````

````python
fig, axes = plot_cross_assay_coverage(
    cyp_overall_coverage,
    title=f"Overall cross-assay, CYP isoform (ChEMBL {CHEMBL_VERSION})",
)
fig.savefig(
    figure_output_path(f"cross_assay_overall_coverage_chembl{CHEMBL_VERSION}.png"),
    dpi=300,
    bbox_inches="tight",
    facecolor=fig.get_facecolor(),
)
plt.show()
````

````python
fig, axes = plot_cross_assay_coverage(
    cyp_coverage,
    title=f"Cross-assay coverage per CYP isoform (ChEMBL {CHEMBL_VERSION})",
)
fig.savefig(
    figure_output_path(f"cross_assay_coverage_chembl{CHEMBL_VERSION}.png"),
    dpi=300,
    bbox_inches="tight",
    facecolor=fig.get_facecolor(),
)
plt.show()
````

## Prepare data for multitask modeling

The `capricho prepare` command filters quality flags and pivots the data into an activity
matrix (compounds × tasks) ready for multitask ML. Like `capricho get`, the underlying
functions can be called directly from Python.

**Option 1** — CLI:
```bash
capricho prepare -i cyp_chembl_data.csv -o cyp_activity_matrix.csv \
    --task-col target_chembl_id --id-columns standard_type \
    --drop-unit-error --drop-undefined-stereo --drop-assay-size
```

**Option 2** — Python API (used below): import `clean_data` and `prepare_multitask_data` directly.

````python
from Capricho.cli.prepare import clean_data, prepare_multitask_data

cleaned = clean_data(
    df,
    drop_flags=[
        DroppingComment.UNIT_ANNOTATION_ERROR.value,
        DroppingComment.UNDEFINED_STEREOCHEMISTRY.value,
        DroppingComment.ASSAY_SIZE_TOO_SMALL.value,
    ],
    deduplicate=True,
)

activity_matrix = prepare_multitask_data(
    df=cleaned,
    task_col="target_chembl_id",
    value_col="pchembl_value_mean",
    compound_col="smiles",
    smiles_col="smiles",
    id_columns=["standard_type"],
)
activity_matrix.head(1)

activity_matrix.to_csv(f"cyp_activity_matrix_chembl{CHEMBL_VERSION}.csv", index=False)
````

````python
from Capricho.chembl.api.downloader import get_target_names_sql

target_ids = [c.split("-")[0] for c in activity_matrix.columns if c.startswith("CHEMBL")]
target_names = get_target_names_sql(target_ids, version=CHEMBL_VERSION)
df["target_name"] = df["target_chembl_id"].map(target_names)
df.columns = [
    target_names[c.split("-")[0]] + c.split("-")[1] if c.startswith("CHEMBL") else c for c in df.columns
]
# df.target_name = df.target_name.apply(lambda x: x.split(",")[0])

df["target_activ_type"] = df["target_name"] + "-" + df["standard_type"]
tasks = df["target_activ_type"].unique().tolist()


_keep = df.target_activ_type.value_counts() >= 100
filtered_tasks = [t for t in tasks if _keep[t]]

result = (
    df.query("target_activ_type in @filtered_tasks")
    .pivot_table(
        index=["connectivity", "smiles"],
        columns="target_activ_type",
        values="pchembl_value_mean",
        aggfunc="first",  # or 'mean' if you have multiple values per combination
    )
    .reset_index()
)
````

### Filter activity sets

After running the cleaning and preparation steps, we remove activity sets with fewer than 100 compounds (see part with query for < 100 compounds below)

````python
from Capricho.chembl.api.downloader import get_target_names_sql
import seaborn as sns

target_ids = [c.split("-")[0] for c in activity_matrix.columns if c.startswith("CHEMBL")]
target_names = get_target_names_sql(target_ids, version=CHEMBL_VERSION)
activity_matrix.columns = [
    target_names[c.split("-")[0]] + " - " + c.split("-")[1] if c.startswith("CHEMBL") else c
    for c in activity_matrix.columns
]
task_cols = [c for c in activity_matrix.columns if c not in ["smiles", "connectivity"]]

n_datapoints = activity_matrix[task_cols].notna().sum()
drop_tasks = n_datapoints[n_datapoints < 100].index.tolist()
filtered_task_cols = [c for c in task_cols if c not in drop_tasks]

activity_matrix.to_csv(f"cyp_activity_matrix_chembl{CHEMBL_VERSION}.csv", index=False)
````

````python
corr_matrix = activity_matrix[filtered_task_cols].corr(method="spearman")
short_names = {
    col: col.replace("Cytochrome P450 ", "CYP ").replace(", mitochondrial", "") for col in corr_matrix.columns
}
corr_matrix = corr_matrix.rename(columns=short_names, index=short_names)

# figsize = (max(10, len(filtered_task_cols) * 0.55), max(8, len(filtered_task_cols) * 0.5))
figsize = (13, 12.0)
fig, ax = plt.subplots(figsize=figsize)
mask = np.triu(np.ones_like(corr_matrix, dtype=bool), k=1)

heatmap = sns.heatmap(
    corr_matrix,
    cmap="coolwarm",
    center=0,
    ax=ax,
    annot=True,
    fmt=".2f",
    mask=mask,
    square=True,
    cbar_kws={"shrink": 0.75, "pad": 0.02, "aspect": 30},
    annot_kws={"fontsize": 10},
)

cbar = heatmap.collections[0].colorbar
cbar.outline.set_edgecolor("black")
cbar.outline.set_linewidth(0.5)
cbar.set_label("Spearman Correlation", labelpad=15, fontsize=12)

plt.setp(ax.get_xticklabels(), rotation=45, ha="right", rotation_mode="anchor", fontsize=10)
plt.setp(ax.get_yticklabels(), rotation=0, fontsize=10)

ax.set_xlabel("")
ax.set_ylabel("")
ax.set_title(f"CYP Isoform Activity Correlation (ChEMBL {CHEMBL_VERSION})", fontsize=16)

fig.tight_layout()
fig.savefig(
    figure_output_path(f"cyp_activity_correlation_matrix_chembl{CHEMBL_VERSION}.png"),
    dpi=300,
    bbox_inches="tight",
    transparent=False,
)
plt.show()
````
